```python
# Exercise 1 & 2: Setup and configuration

import pandas as pd
import logging

# Constants (simulate config.py)
CSV_PATH = "sample_data.csv"
THRESHOLD_AGE = 18
LOG_FILE = "app.log"

# Configure logging (simulate logging setup in main)
logging.basicConfig(
    filename=LOG_FILE,
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
)

print("✅ Configuration and logging set up successfully.")
```

```python
# Exercise 3: Add exception handling and logging

def load_data(path: str = CSV_PATH) -> pd.DataFrame:
    """
    Load a CSV file into a Pandas DataFrame.

    Args:
        path (str): Path to the CSV file.

    Returns:
        pd.DataFrame: The loaded dataset.

    Raises:
        FileNotFoundError: If the file cannot be found.
    """
    try:
        data = pd.read_csv(path)
        logging.info(f"Successfully loaded data from {path}")
        return data
    except FileNotFoundError:
        logging.error(f"File not found at path: {path}")
        raise FileNotFoundError(f"Could not find file at: {path}")
```

```python
# Exercise 4: Add modularized cleaning logic

def clean_data(df: pd.DataFrame) -> pd.DataFrame:
    """
    Clean and preprocess the dataset.

    Operations:
    - Drop rows with missing values
    - Filter rows where age >= THRESHOLD_AGE (from constants)

    Args:
        df (pd.DataFrame): Raw dataset

    Returns:
        pd.DataFrame: Cleaned dataset
    """
    logging.info("Starting data cleaning process.")
    initial_shape = df.shape

    df = df.dropna()
    logging.info(f"Removed missing values. Shape: {df.shape}")

    if "age" in df.columns:
        df = df[df["age"] >= THRESHOLD_AGE]
        logging.info(f"Filtered rows with age >= {THRESHOLD_AGE}. Shape: {df.shape}")
    else:
        logging.warning("'age' column not found — no filtering applied.")

    logging.info(f"Data cleaned. Original shape {initial_shape} -> {df.shape}")
    return df
```

```python
def main():
    """
    Main workflow to load, clean, and log dataset processing.
    """
    print("🚀 Starting TP-2 Data Preprocessing Pipeline...\n")
    logging.info("Pipeline started.")

    try:
        # Load
        df = load_data()
        print("📂 Loaded Data:")
        display(df.head())

        # Clean
        cleaned = clean_data(df)
        print("\n🧹 Cleaned Data:")
        display(cleaned.head())

        logging.info(f"Pipeline completed. Final dataset shape: {cleaned.shape}")
        print("\n✅ Pipeline completed successfully!")

    except Exception as e:
        logging.error(f"Pipeline error: {e}")
        print("❌ Error occurred — check tp2_log.log for details.")


# Run pipeline
main()
```
